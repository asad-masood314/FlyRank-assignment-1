# Lane 2 — Refresh / Content Opportunity Scoring

## Abstract

This study investigates whether historical search and content signals can help prioritize webpages for content review. I constructed page-level historical features and defined a future search-performance decline proxy as the prediction target. I compared a simple baseline with a Random Forest model and evaluated the model under the original and client-grouped validation designs. The model measured a Precision@50 of 0.90 under the original split and 0.52 under client-grouped validation. The results provide directional decision-support for prioritizing page review, but they do not prove causal refresh impact or predict Google's ranking algorithm.

## 2. Data

The analysis uses the FlyRank internship warehouse accessed through DuckDB. The main source is the daily content-performance data, with content information used where needed.

Historical observations are used to build the model features, while a later observation window is used only to define the future outcome proxy. Future-window performance information is excluded from the model features to reduce leakage.

The public research does not include client names, domains, URLs, private queries, credentials, or raw data exports.

### Methodology
Which pages should be reviewed first for refresh, expansion, protection, pruning, or monitoring based on historical search and content signals?

### Feature Engineering

The initial Lane 2 feature set uses historical average impressions, historical average clicks, and historical average search position. These features are intended to represent information available before the prediction decision.

Client and content identifiers are used for grouping and joining, not as predictive features.

```python
features = [
    "avg_impressions",
    "avg_clicks",
    "avg_position"
]

print("Features used:")
for feature in features:
    print("-", feature)
```

## 3. Target Definition

The target is a proxy for future search-performance decline. A page is labelled positive when its future average impressions are at least 20% lower than its historical average impressions.

This is an operational proxy for future decline and not a direct measurement of whether a page actually needs a refresh.

### Baseline

The baseline is a simple rule-based approach that provides a transparent reference point for evaluating the learned model. The baseline is intentionally simple so that any improvement from the model can be compared against an understandable starting point.

```python
baseline_precision_at_50 = 0.90

print("Week 5 reported Precision@50:", baseline_precision_at_50)
```

### Model

I used a Random Forest classifier to learn relationships between historical page signals and the future-decline proxy. The model produces a score that can be used to rank pages by predicted likelihood of the defined future outcome.

### Ranked Recommendations

The final output ranks pages by their model score and assigns an action for human review. Higher-scoring pages receive stronger review priority, while lower-scoring pages remain in monitoring.

These recommendations are decision-support and do not automatically determine what content should be changed.

### Limitations

The target is a proxy for future search-performance decline rather than a direct measure of whether a page should be refreshed. The evaluation covers the tested warehouse slice and may not generalize to all clients or websites.

The difference between the original Precision@50 of 0.90 and the client-grouped Precision@50 of 0.52 shows that validation design materially affects measured performance.

The findings are observed and measured within the tested setup and should be treated as directional decision-support. They do not prove that refreshing a recommended page will cause performance to improve, and they do not predict Google's ranking algorithm.
