```python
import pandas as pd

df = pd.read_csv("../outputs/lane2_training_data.csv")

print(df.shape)
df.head()
```

```python
features = [
    "avg_impressions",
    "avg_clicks",
    "avg_position"
]

target = "future_decline"

X = df[features]
y = df[target]

print("Features:", features)
print("Target:", target)
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Test rows:", len(X_test))
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)

print("Model trained successfully.")
```

```python
test_scores = model.predict_proba(X_test)[:, 1]

print(test_scores[:10])
```

```python
results = pd.DataFrame({
    "actual": y_test.to_numpy(),
    "score": test_scores
})

results = results.sort_values(
    "score",
    ascending=False
).reset_index(drop=True)

results.head(10)
```

```python
k = 10

precision_at_10 = results.head(k)["actual"].mean()

print("Precision@10:", precision_at_10)
```

```python
k = 20

precision_at_20 = results.head(k)["actual"].mean()

print("Precision@20:", precision_at_20)
```

```python
k = 50

precision_at_50 = results.head(k)["actual"].mean()

print("Precision@50:", precision_at_50)
```
